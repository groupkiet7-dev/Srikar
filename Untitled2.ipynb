```python
# ================================================================
# BRIDGE LOAD DISTRIBUTION CALCULATOR (A2)
# B.Tech Mathematics Mini Project
# ================================================================
#
# Mathematical concept:
#       AX = B
#       Unique solution
#       Matrix inverse
#
# Environment:
#       Google Colab + Python
#       NumPy + SymPy + Pandas + Matplotlib
#
# ================================================================

# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import numpy as np
import sympy as sp
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display, Markdown


# ================================================================
# 2. PROJECT TITLE
# ================================================================

print("=" * 75)
print("              BRIDGE LOAD DISTRIBUTION CALCULATOR")
print("              B.Tech Mathematics Mini Project - A2")
print("=" * 75)

print("""
Problem Statement:
Three supports hold a bridge. Loads act at different points.
The reaction forces at the three supports are calculated by
solving a system of three linear equations.

Mathematical model:

                    AX = B

For a unique solution:

                    det(A) != 0

Using matrix inverse:

                    X = A^(-1) B

A truck is then moved to a different position and all three
support reactions are recalculated.
""")


# ================================================================
# 3. MODELING ASSUMPTION
# ================================================================

print("=" * 75)
print("MODELING ASSUMPTION")
print("=" * 75)

print("""
Important assumption:

For three unknown support reactions R1, R2 and R3, the two basic
static equilibrium equations alone are not sufficient to determine
three unknowns.

Therefore, this educational model uses the following additional
compatibility assumption:

                    R1 = R2

or:

                    R1 - R2 = 0

This provides the third independent equation required for a 3 x 3
system.

This is a simplified mathematical model for demonstrating matrix
methods. It is NOT a real structural-engineering design formula.
""")


# ================================================================
# 4. USER INPUTS
# ================================================================
#
# You can change these values to test different bridge situations.
# Units:
#   Distance = metres
#   Force    = kN
# ================================================================

# ------------------------------------------------
# SUPPORT POSITIONS
# ------------------------------------------------

support_positions = np.array([
    0.0,      # Support 1
    10.0,     # Support 2
    20.0      # Support 3
])


# ------------------------------------------------
# INITIAL LOADS
# ------------------------------------------------

initial_loads = [

    {
        "name": "Load 1",
        "position": 5.0,
        "force": 100.0
    },

    {
        "name": "Truck",
        "position": 12.0,
        "force": 150.0
    },

    {
        "name": "Load 3",
        "position": 17.0,
        "force": 80.0
    }

]


# ------------------------------------------------
# CHANGED LOADS
# ------------------------------------------------
#
# Truck moves from 12 m to 15 m.
# Its force remains 150 kN.
# ------------------------------------------------

changed_loads = [

    {
        "name": "Load 1",
        "position": 5.0,
        "force": 100.0
    },

    {
        "name": "Truck",
        "position": 15.0,
        "force": 150.0
    },

    {
        "name": "Load 3",
        "position": 17.0,
        "force": 80.0
    }

]


# ================================================================
# 5. DISPLAY INPUT DATA
# ================================================================

print("=" * 75)
print("INPUT DATA")
print("=" * 75)

print("\nSupport Positions:")

for i, position in enumerate(support_positions, start=1):
    print(
        f"Support {i}: {position:.2f} m"
    )


print("\nInitial Loads:")

for load in initial_loads:

    print(
        f"{load['name']}: "
        f"{load['force']:.2f} kN "
        f"at {load['position']:.2f} m"
    )


print("\nChanged Loads:")

for load in changed_loads:

    print(
        f"{load['name']}: "
        f"{load['force']:.2f} kN "
        f"at {load['position']:.2f} m"
    )


# ================================================================
# 6. INPUT VALIDATION
# ================================================================

def validate_inputs(support_positions, loads):

    # Exactly three supports are required
    if len(support_positions) != 3:
        raise ValueError(
            "Exactly 3 supports are required."
        )

    # Support positions must be different
    if len(set(support_positions)) != 3:
        raise ValueError(
            "Support positions must be different."
        )

    # Support positions must be ordered
    if not np.all(
        np.diff(support_positions) > 0
    ):
        raise ValueError(
            "Support positions must be in increasing order."
        )

    # At least one load
    if len(loads) == 0:
        raise ValueError(
            "At least one load is required."
        )

    # Check every load
    for load in loads:

        if "position" not in load:
            raise ValueError(
                "Every load needs a position."
            )

        if "force" not in load:
            raise ValueError(
                "Every load needs a force."
            )

        if load["force"] < 0:
            raise ValueError(
                "Load force cannot be negative."
            )


validate_inputs(
    support_positions,
    initial_loads
)

validate_inputs(
    support_positions,
    changed_loads
)

print("\nInput validation successful.")


# ================================================================
# 7. CONSTRUCT AX = B
# ================================================================

def build_system(
    support_positions,
    loads
):
    """
    Construct the linear system AX = B.

    Unknown vector:

             | R1 |
        X =  | R2 |
             | R3 |

    Equation 1:
        R1 + R2 + R3 = Total Load

    Equation 2:
        x1R1 + x2R2 + x3R3 = Total Moment

    Equation 3:
        R1 - R2 = 0
    """

    # ------------------------------------------------------------
    # Support positions
    # ------------------------------------------------------------

    x1 = support_positions[0]
    x2 = support_positions[1]
    x3 = support_positions[2]


    # ------------------------------------------------------------
    # Total vertical load
    # ------------------------------------------------------------

    total_load = sum(
        load["force"]
        for load in loads
    )


    # ------------------------------------------------------------
    # Total moment about x = 0
    # ------------------------------------------------------------

    total_moment = sum(
        load["force"] * load["position"]
        for load in loads
    )


    # ------------------------------------------------------------
    # Matrix A
    # ------------------------------------------------------------

    A = np.array([

        # Equation 1:
        # R1 + R2 + R3 = Total Load
        [1, 1, 1],

        # Equation 2:
        # x1R1 + x2R2 + x3R3 = Total Moment
        [x1, x2, x3],

        # Equation 3:
        # R1 - R2 = 0
        [1, -1, 0]

    ], dtype=float)


    # ------------------------------------------------------------
    # Vector B
    # ------------------------------------------------------------

    B = np.array([

        total_load,

        total_moment,

        0

    ], dtype=float)


    return A, B


# Build initial system
A_initial, B_initial = build_system(
    support_positions,
    initial_loads
)


# Build changed system
A_changed, B_changed = build_system(
    support_positions,
    changed_loads
)


# ================================================================
# 8. DISPLAY AX = B
# ================================================================

print("\n" + "=" * 75)
print("MATHEMATICAL MODEL: INITIAL CONDITION")
print("=" * 75)

print("\nMatrix A:")
print(A_initial)

print("\nVector B:")
print(B_initial)

print("\nTherefore:")

print("""
       AX = B

       where

       X = [R1  R2  R3]^T
""")


# ================================================================
# 9. DISPLAY EQUATIONS USING SYMPY
# ================================================================

print("=" * 75)
print("THREE MATHEMATICAL EQUATIONS")
print("=" * 75)

A_sym = sp.Matrix(A_initial)
B_sym = sp.Matrix(B_initial)

R1, R2, R3 = sp.symbols(
    "R1 R2 R3"
)

X_sym = sp.Matrix([
    R1,
    R2,
    R3
])

print("\nUnknown vector X:")

sp.pprint(X_sym)


print("\nEquations:")

equations = A_sym * X_sym

for i in range(3):

    equation = sp.Eq(
        equations[i],
        B_sym[i]
    )

    print(
        f"\nEquation {i + 1}:"
    )

    sp.pprint(equation)


# ================================================================
# 10. UNIQUE SOLUTION CHECK
# ================================================================

print("\n" + "=" * 75)
print("UNIQUE SOLUTION CHECK")
print("=" * 75)

det_A = np.linalg.det(
    A_initial
)

rank_A = np.linalg.matrix_rank(
    A_initial
)

print(
    f"\ndet(A) = {det_A:.6f}"
)

print(
    f"rank(A) = {rank_A}"
)


if (
    abs(det_A) > 1e-10
    and rank_A == 3
):

    print(
        "\n✓ det(A) is not zero."
    )

    print(
        "✓ Matrix A is invertible."
    )

    print(
        "✓ The system has a UNIQUE solution."
    )

else:

    raise ValueError(
        "The system does not have a unique solution."
    )


# ================================================================
# 11. MATRIX INVERSE
# ================================================================

A_inverse = np.linalg.inv(
    A_initial
)

print("\n" + "=" * 75)
print("MATRIX INVERSE")
print("=" * 75)

print("\nA^(-1) =")

print(A_inverse)


# ================================================================
# 12. SOLVE USING X = A^(-1)B
# ================================================================

initial_reactions = (
    A_inverse @ B_initial
)


# Solve changed condition
changed_inverse = np.linalg.inv(
    A_changed
)

changed_reactions = (
    changed_inverse @ B_changed
)


print("\n" + "=" * 75)
print("MATRIX INVERSE SOLUTION")
print("=" * 75)

print("""
Starting with:

        AX = B

Multiply both sides by A^(-1):

        A^(-1)AX = A^(-1)B

Since:

        A^(-1)A = I

we get:

        X = A^(-1)B
""")


print("\nInitial reaction vector X:")

print(initial_reactions)


print("\nInitial support reactions:")

print(
    f"R1 = {initial_reactions[0]:.3f} kN"
)

print(
    f"R2 = {initial_reactions[1]:.3f} kN"
)

print(
    f"R3 = {initial_reactions[2]:.3f} kN"
)


# ================================================================
# 13. VERIFY SOLUTION
# ================================================================

print("\n" + "=" * 75)
print("SOLUTION VERIFICATION")
print("=" * 75)

calculated_B = (
    A_initial @ initial_reactions
)

residual = (
    calculated_B - B_initial
)

print("\nOriginal B:")

print(B_initial)


print("\nCalculated AX:")

print(calculated_B)


print("\nResidual AX - B:")

print(residual)


maximum_error = np.max(
    np.abs(residual)
)

print(
    f"\nMaximum absolute error = "
    f"{maximum_error:.3e}"
)


if maximum_error < 1e-10:

    print(
        "\n✓ AX = B verified successfully."
    )

else:

    print(
        "\n⚠ Check numerical accuracy."
    )


# ================================================================
# 14. INITIAL VS CHANGED RESULTS
# ================================================================

reaction_change = (
    changed_reactions -
    initial_reactions
)


results = pd.DataFrame({

    "Support": [
        "Support 1",
        "Support 2",
        "Support 3"
    ],

    "Position (m)": (
        support_positions
    ),

    "Initial Reaction (kN)": (
        initial_reactions
    ),

    "After Truck Moves (kN)": (
        changed_reactions
    ),

    "Change (kN)": (
        reaction_change
    )

})


results = results.round(3)


print("\n" + "=" * 75)
print("FINAL RESULTS")
print("=" * 75)

display(results)


# ================================================================
# 15. TEXT INTERPRETATION
# ================================================================

print("\n" + "=" * 75)
print("INTERPRETATION")
print("=" * 75)

print("""
The truck initially acts at 12 m.

The truck is then moved to 15 m.

The total load remains the same, but the total moment changes.
Therefore, the vector B changes.

Since:

        X = A^(-1)B

the support reaction vector X also changes.
""")

for i in range(3):

    print(
        f"Support {i + 1}: "
        f"{initial_reactions[i]:.2f} kN "
        f"→ "
        f"{changed_reactions[i]:.2f} kN "
        f"({reaction_change[i]:+.2f} kN)"
    )


# ================================================================
# 16. CHART 1 — BRIDGE LOAD DIAGRAM
# ================================================================

fig, ax = plt.subplots(
    figsize=(12, 5)
)


# Bridge deck
ax.plot(
    support_positions,
    [0, 0, 0],
    linewidth=8,
    solid_capstyle="round"
)


# Supports
for i, position in enumerate(
    support_positions
):

    ax.scatter(
        position,
        -0.25,
        marker="^",
        s=300
    )

    ax.text(
        position,
        -0.65,
        f"R{i + 1}",
        ha="center",
        fontsize=12,
        fontweight="bold"
    )


# Loads
for load in initial_loads:

    ax.annotate(

        f"{load['name']}\n"
        f"{load['force']:.0f} kN",

        xy=(
            load["position"],
            0
        ),

        xytext=(
            load["position"],
            1.4
        ),

        ha="center",

        arrowprops=dict(
            arrowstyle="->",
            linewidth=2
        )

    )


ax.set_title(
    "Bridge Load Distribution Model",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlabel(
    "Position along bridge (m)"
)

ax.set_ylabel(
    "Schematic position"
)

ax.set_xlim(
    -2,
    22
)

ax.set_ylim(
    -1.2,
    2.3
)

ax.grid(
    True,
    alpha=0.3
)

plt.show()


# ================================================================
# 17. CHART 2 — INITIAL VS CHANGED REACTIONS
# ================================================================

support_numbers = np.arange(3)

width = 0.35

fig, ax = plt.subplots(
    figsize=(10, 6)
)


ax.bar(
    support_numbers - width / 2,
    initial_reactions,
    width,
    label="Before truck moves"
)


ax.bar(
    support_numbers + width / 2,
    changed_reactions,
    width,
    label="After truck moves"
)


ax.set_title(
    "Support Reactions Before and After Truck Movement",
    fontsize=15,
    fontweight="bold"
)

ax.set_xlabel(
    "Support"
)

ax.set_ylabel(
    "Reaction Force (kN)"
)

ax.set_xticks(
    support_numbers
)

ax.set_xticklabels([
    "Support 1",
    "Support 2",
    "Support 3"
])

ax.legend()

ax.grid(
    axis="y",
    alpha=0.3
)

plt.show()


# ================================================================
# 18. CHART 3 — CHANGE IN REACTIONS
# ================================================================

fig, ax = plt.subplots(
    figsize=(10, 5)
)


bars = ax.bar(
    [
        "Support 1",
        "Support 2",
        "Support 3"
    ],
    reaction_change
)


ax.axhline(
    0,
    linewidth=1
)


ax.set_title(
    "Change in Support Reactions",
    fontsize=15,
    fontweight="bold"
)

ax.set_xlabel(
    "Support"
)

ax.set_ylabel(
    "Change in Reaction (kN)"
)

ax.grid(
    axis="y",
    alpha=0.3
)


# Add numerical values
for bar, value in zip(
    bars,
    reaction_change
):

    height = bar.get_height()

    ax.text(

        bar.get_x()
        + bar.get_width() / 2,

        height,

        f"{value:+.2f}",

        ha="center",

        va=(
            "bottom"
            if value >= 0
            else "top"
        )

    )


plt.show()


# ================================================================
# 19. TRUCK MOVEMENT SIMULATION
# ================================================================

print("\n" + "=" * 75)
print("TRUCK MOVEMENT SIMULATION")
print("=" * 75)


# Fixed loads
fixed_loads = [

    {
        "name": "Fixed Load 1",
        "position": 5.0,
        "force": 100.0
    },

    {
        "name": "Fixed Load 2",
        "position": 17.0,
        "force": 80.0
    }

]


# Truck force
truck_force = 150.0


# Truck moves between 3 m and 18 m
truck_positions = np.linspace(
    3,
    18,
    50
)


# Store calculated reactions
reaction_history = []


# Calculate reactions at every truck position
for truck_position in truck_positions:

    current_loads = (
        fixed_loads
        + [
            {
                "name": "Moving Truck",
                "position": truck_position,
                "force": truck_force
            }
        ]
    )


    A_current, B_current = (
        build_system(
            support_positions,
            current_loads
        )
    )


    # Check invertibility
    if abs(
        np.linalg.det(A_current)
    ) < 1e-10:

        raise ValueError(
            "Matrix became singular."
        )


    # Calculate reaction vector
    current_reactions = (
        np.linalg.inv(A_current)
        @ B_current
    )


    reaction_history.append(
        current_reactions
    )


# Convert to NumPy array
reaction_history = np.array(
    reaction_history
)


print(
    "Truck movement simulation completed."
)


# ================================================================
# 20. CHART 4 — REACTION VS TRUCK POSITION
# ================================================================

fig, ax = plt.subplots(
    figsize=(11, 6)
)


for i in range(3):

    ax.plot(

        truck_positions,

        reaction_history[:, i],

        linewidth=2.5,

        label=f"Support {i + 1}"

    )


ax.set_title(
    "Support Reactions as the Truck Moves",
    fontsize=16,
    fontweight="bold"
)

ax.set_xlabel(
    "Truck Position (m)"
)

ax.set_ylabel(
    "Reaction Force (kN)"
)

ax.legend()

ax.grid(
    True,
    alpha=0.3
)

plt.show()


# ================================================================
# 21. FINAL PROJECT SUMMARY
# ================================================================

print("\n" + "=" * 75)
print("                    PROJECT COMPLETED")
print("=" * 75)

print("""
MAIN MATHEMATICAL STEPS
-----------------------

1. Define bridge supports and loads.

2. Calculate total vertical load:

       Total Load = ΣP

3. Calculate total moment:

       Total Moment = Σ(P × x)

4. Construct:

       AX = B

5. Check:

       det(A) != 0

   Therefore A has an inverse and the solution is unique.

6. Calculate:

       X = A^(-1)B

7. Obtain:

       R1, R2, R3

8. Move the truck.

9. Recalculate B.

10. Calculate the new reaction vector.

11. Compare the old and new reactions.

12. Plot the results.

CONCLUSION
----------

Moving a load changes the moment acting on the bridge.
Consequently, the calculated support reactions change.

This demonstrates the application of matrix methods to a
simplified structural-engineering load-distribution problem.
""")

print("=" * 75)
print("END OF PROGRAM")
print("=" * 75)
```
