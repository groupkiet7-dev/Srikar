```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as w
from IPython.display import display, clear_output

# Bridge supports
S = [0, 10, 20]

# Loads: force (kN), position (m)
LOADS = [(100, 5), (150, 12), (80, 17)]

def calculate(truck):
    loads = [(100, 5), (150, truck), (80, 17)]

    A = np.array([
        [1, 1, 1],
        S,
        [1, -1, 0]
    ], dtype=float)

    B = np.array([
        sum(f for f, p in loads),
        sum(f*p for f, p in loads),
        0
    ])

    X = np.linalg.solve(A, B)
    return A, B, X

slider = w.FloatSlider(
    value=12,
    min=3,
    max=18,
    step=0.5,
    description="🚚 Truck:"
)

out = w.Output()

def show(truck):
    with out:
        clear_output(wait=True)

        A, B, X = calculate(truck)

        print("🚧 BRIDGE LOAD DISTRIBUTION CALCULATOR")
        print("=" * 45)

        print(f"\nTruck Position: {truck:.1f} m")
        print(f"R1 = {X[0]:.2f} kN")
        print(f"R2 = {X[1]:.2f} kN")
        print(f"R3 = {X[2]:.2f} kN")

        print("\nMatrix A:")
        print(A)

        print("\nVector B:")
        print(B)

        print("\nSolution X = A⁻¹B:")
        print(np.round(X, 2))

        # Bridge diagram
        plt.figure(figsize=(10, 4))
        plt.plot(S, [0, 0, 0], linewidth=8)

        for i, x in enumerate(S):
            plt.scatter(x, -0.1, marker="^", s=300)
            plt.text(
                x, -0.45,
                f"R{i+1}={X[i]:.1f} kN",
                ha="center"
            )

        for force, pos in [(100,5), (150,truck), (80,17)]:
            plt.annotate(
                f"{force} kN",
                (pos, 0),
                (pos, 1.5),
                ha="center",
                arrowprops=dict(arrowstyle="->")
            )

        plt.title("Bridge Load Distribution")
        plt.xlabel("Position (m)")
        plt.yticks([])
        plt.grid(axis="x", linestyle=":")
        plt.show()

slider.observe(lambda x: show(x["new"]), names="value")

display(slider, out)
show(12)
```
